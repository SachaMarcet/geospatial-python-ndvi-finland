```python
import numpy as np
import matplotlib.pyplot as plt
import rasterio
```

```python
with rasterio.open("data/B04.tiff") as src:
    b04 = src.read(1)

with rasterio.open("data/B08.tiff") as src:
    b08 = src.read(1)
```

```python
def read_band(path):
    with rasterio.open(path) as src:
        return src.read(1)

def compute_ndvi(nir, red):
    nir = nir.astype("float32")
    red = red.astype("float32")
    return (nir - red) / (nir + red + 1e-10)

red = read_band("data/B04.tiff")
nir = read_band("data/B08.tiff")

ndvi = compute_ndvi(nir, red)
```

```python


plt.imshow(ndvi, cmap="RdYlGn")
plt.colorbar()
plt.title("NDVI")
plt.show()
```
